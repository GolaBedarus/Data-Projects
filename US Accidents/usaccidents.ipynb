When do accidents most commonly occur? 

```python
df.Start_Time
```

```python
df.Start_Time = pd.to_datetime(df.Start_Time, format="%Y-%m-%d %H:%M:%S", errors="coerce")
```

```python
df.Start_Time[0]
```

```python
sns.distplot(df.Start_Time.dt.hour, bins=24, kde= False, norm_hist= True)
```

```python
sns.distplot(df.Start_Time.dt.dayofweek, bins=7, kde= False, norm_hist= True)
```

Is the distribution of accidents by hour the same on weekends as on weekdays?

```python
sundays_start_time = df.Start_Time[df.Start_Time.dt.dayofweek == 6]
sns.distplot(sundays_start_time.dt.hour, bins= 24, kde= False, norm_hist= True)
```

```python
monday_start_time = df.Start_Time[df.Start_Time.dt.dayofweek == 0]
sns.distplot(monday_start_time.dt.hour, bins= 24, kde= False, norm_hist= True)
```

What about the months? Is there any month in which there are more accidents?

```python
sns.distplot(df.Start_Time.dt.month, bins= 12, kde = False, norm_hist = True)
```

```python
df_2019 = df[df.Start_Time.dt.year == 2019]
sns.distplot(df_2019.Start_Time.dt.month, bins= 12, kde = False, norm_hist= True)
```

## DATA AND INSIGHTS

1- New York accident number is surprisingly low.

2- Less than 8% of cities have more than 1000 yearly accidents.

3- 1023 cities have reported having 1 accident (seems off and can't report any useful insight).

4- The number of accidents per city increases/decreases exponentially.

5- A high percentage of accidents occur between 7 to 9 AM (Probably people in a hurry to get to work). Next Highest percentage is from 3 to 6 PM (maybe peoble returning home from work).

7- On weekends, the number of accidents is lower.

8- On Sundays the peak occurs between 10 AM and 15 PM.

9- Much data is missing from 2016.
